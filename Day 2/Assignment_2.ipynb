```python
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer

knowledge_base = [
    "How to reset your forgotten password from the account settings.",
    "Updating payment methods, credit cards, and billing addresses.",
    "Troubleshooting login issues and two-factor authentication errors.",
    "Managing account settings, profile information, and preferences.",
    "How to request a refund for recent charges on your subscription.",
    "Unlocking a locked account after multiple failed login attempts.",
    "Viewing past invoices and downloading annual billing statements.",
    "Changing your registered email address or account username.",
    "Upgrading or downgrading your monthly membership plan.",
    "Contacting customer support for urgent account security concerns."
]

model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = model.encode(knowledge_base, convert_to_numpy=True).astype(np.float32)

print("--- TASK 1 OUTPUT ---")
print(f"Knowledge Base Size: {len(knowledge_base)} sentences")
print(f"Embedding Matrix Shape: {embeddings.shape}\n")


dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
faiss.normalize_L2(embeddings)
index.add(embeddings)

print("--- TASK 2 OUTPUT ---")
print(f"Total vectors stored in FAISS index: {index.ntotal}\n")

def search_faiss(query: str, k: int = 3):
    query_vector = model.encode([query], convert_to_numpy=True).astype(np.float32)

    faiss.normalize_L2(query_vector)
        distances, indices = index.search(query_vector, k)

    print(f"Query: '{query}'")
    print(f"{'Rank':<6} | {'Distance/Score':<15} | {'Matched Sentence'}")
    print("-" * 75)

    for rank, (dist, idx) in enumerate(zip(distances[0], indices[0]), start=1):
        matched_text = knowledge_base[idx]
        print(f"{rank:<6} | {dist:<15.4f} | {matched_text}")
    print("\n" + "="*75 + "\n")

test_queries = [
    "I forgot my password and cannot sign in",
    "Where can I find my past receipts and payment history?",
    "I want to change my email address"
]

print("--- TASK 3 OUTPUT: TESTING 3 QUERIES ---")
for query in test_queries:
    search_faiss(query, k=3)
```

```python
print("=== FAISS Semantic Search Engine CLI ===")
print("Type your query to search the knowledge base. Type 'exit' to quit.\n")

while True:
    user_query = input("Enter query: ").strip()

    if user_query.lower() == "exit":
        print("Exiting search engine. Goodbye!")
        break

    if not user_query:
        continue

    search_faiss(user_query, k=3)
```
